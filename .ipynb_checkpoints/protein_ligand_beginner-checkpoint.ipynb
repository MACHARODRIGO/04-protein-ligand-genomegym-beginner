## 📂 1. Protein Structure with Gemmi

### 1.1 Download example protein (1CRN)
Small plant protein (Crambin, 46 residues) from RCSB PDB.


```python
import os
import gemmi
import urllib.request


# Definir directorio y archivo
DATA_DIR = "data"
FILENAME = "1CRN.pdb"
FILEPATH = os.path.join(DATA_DIR, FILENAME)

# Crear directorio si no existe
os.makedirs(DATA_DIR, exist_ok=True)

# Crear el directorio "data" si no existe
os.makedirs("data", exist_ok=True)

# Descargar solo si no existe
url = "https://files.rcsb.org/download/1CRN.pdb"
if not os.path.exists(FILEPATH):
    urllib.request.urlretrieve(url, FILEPATH)
    print(f"Downloaded {FILENAME}")
```

### 1.2 Load structure with Gemmi
Read the PDB file and access the first model.

```python
structure = gemmi.read_structure(FILEPATH)
model = structure[0]

print(f"Model contains {len(model)} chains.")
for chain in model:
    print(f"Chain {chain.name}: {len(chain)} residues")
```

### 1.3 Explore chains and residues
Iterate through chains and count residues.

```python
print("\nFirst 5 residues:")
for res in model[0][:5]:
    print(res.name, res.seqid.num)
```

### 1.4 Inspect first atoms
List first residues and coordinates of the first atom.

```python
atom = model[0][0][0]
print("\nFirst atom:", atom.name, atom.pos)
```

## ⚗️ 2. Small Molecule with RDKit

### 2.1 Build molecule from SMILES
Use ethanol (CCO) as a model.
Check that the SMILES string corresponds to a valid molecule.

```python
from rdkit import Chem
from rdkit.Chem import Draw, rdMolDescriptors, Descriptors
from IPython.display import display

smiles = "CCO"  # ethanol
mol = Chem.MolFromSmiles(smiles)

if mol is None:
    raise ValueError("❌ Invalid SMILES string: molecule could not be loaded.")

print("\n✅ Molecule loaded successfully")

```

### 2.2 Analyze Molecule
In this step we will:  
- Calculate its molecular formula and weight  
- Generate and save a 2D image (`ethanol.png`)

```python
# Crear directorio de Resultados si no existe
os.makedirs("results", exist_ok=True)

formula = rdMolDescriptors.CalcMolFormula(mol)
weight = Descriptors.MolWt(mol)
print(f"Molecular formula: {formula}")
print(f"Molecular weight: {weight:.2f} Da")

# Save molecule image as PNG
img = Draw.MolToImage(mol, size=(250,250))
img.save("results/ethanol.png")
print("2D structure image saved as ethanol.png")

# Display inline (works in Jupyter/Colab)
img
```
